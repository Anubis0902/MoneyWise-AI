```python
from langchain_ollama import ChatOllama
from langchain_core.output_parsers import PydanticOutputParser
from langchain_core.prompts import PromptTemplate
from dotenv import load_dotenv
from pydantic import BaseModel , Field
```

```python
load_dotenv ()
model = ChatOllama(model = 'llama3.2:latest')
```

```python
class expense(BaseModel):
    title : str = Field(description='title of that exepnese , eg :- hotel dinner , arcade bowling etc')
    amount : int = Field(description='The amount spent on buying or doing something')
    category : str = Field(description='The category in which the amount is spent you have to guess it and figure it out on you own , for example :- food , education , gift , luxury leisure , etc ')
    mode : str = Field(description='this will be the mode of expense , cash , online , credit card , debit card etc , note if the user does not provide any mode of transaction we will consider it as online ')
```

```python
parser = PydanticOutputParser(pydantic_object= expense)
```

```python
template = PromptTemplate(
    template= "extract title , amount , category , mode from the given expense provided by the user {user_expense} \n {format_instructions}",
    input_variables=['user_expense'],
    partial_variables= {'format_instructions' : parser.get_format_instructions()}
)
```

```python
chain = template | model | parser
final_result = chain.invoke({'user_expense' : '500rs of wheat taken in cash '})
```
